**1. Make Gamma Correction according to the following instructions**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(' Gamma Correction pada citra ')
print('----------------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

img = cv2.imread('/content/drive/MyDrive/Images/female.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

normalized = img / 255.0

gamma_corrected = np.power(normalized, 1.0/gamma)

gamma_corrected = np.uint8(gamma_corrected * 255)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img)
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gamma_corrected)
plt.axis('off')

plt.show()
```

```python
print(' Gamma Correction pada citra (manual loop)')
print('----------------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

img = cv2.imread('/content/drive/MyDrive/Images/female.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gamma_corrected = img.copy()

for y in range(img.shape[0]):
    for x in range(img.shape[1]):
        for c in range(img.shape[2]):
            pixel = img[y, x, c] / 255.0
            new_pixel = 255.0 * (pixel ** (1.0/gamma))
            gamma_corrected[y, x, c] = int(new_pixel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('Citra Asli')
plt.imshow(img)
plt.axis('off')

plt.subplot(1,2,2)
plt.title(f'Gamma Correction (γ={gamma})')
plt.imshow(gamma_corrected)
plt.axis('off')

plt.show()
```

**2. Create Image Depth Simulation**

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

bit_depth = 2
level = 255 / (pow(2, bit_depth) - 1)

original = cv.imread('/content/drive/MyDrive/Images/female.png', cv.IMREAD_GRAYSCALE)

depth_image = np.zeros(original.shape, dtype=np.uint8)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        pixel = original[y, x]
        new_pixel = round(pixel / level) * level
        depth_image[y, x] = np.uint8(new_pixel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('Original (8-bit)')
plt.imshow(original, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title(f'Quantized ({bit_depth}-bit)')
plt.imshow(depth_image, cmap='gray')
plt.axis('off')

plt.show()
```

**3. Create an Average Denoising module**

```python
import cv2 as cv
import numpy as np
import glob
import matplotlib.pyplot as plt
from math import log10, sqrt

def PSNR(img1, img2):
    mse = np.mean((img1 - img2) ** 2)
    if mse == 0:
        return 100
    max_pixel = 255.0
    psnr = 20 * log10(max_pixel / sqrt(mse))
    return psnr

original = cv.imread('/content/drive/MyDrive/Images/galaxy.jpg')
original = cv.cvtColor(original, cv.COLOR_BGR2RGB)

cv_img = []
for img in glob.glob('/content/drive/MyDrive/Images/noises/*.jpg'):
    n = cv.imread(img)
    n = cv.cvtColor(n, cv.COLOR_BGR2RGB)
    cv_img.append(n)

cv_img = np.array(cv_img)

num_images_list = [10, 20, 40, 80, 100]

results = []

for num in num_images_list:
    subset = cv_img[:num]
    avg_img = np.mean(subset, axis=0).astype(np.uint8)

    psnr_value = PSNR(original, avg_img)
    results.append((num, avg_img, psnr_value))

    print(f"Jumlah citra: {num}, PSNR: {psnr_value:.2f} dB")

plt.figure(figsize=(13,6))
for i, (num, avg_img, psnr_value) in enumerate(results):
    plt.subplot(2,3,i+1)
    plt.imshow(avg_img)
    plt.title(f"{num} Images\nPSNR={psnr_value:.2f} dB")
    plt.axis('off')

plt.subplot(2,3,6)
plt.imshow(original)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

The more images used for averaging, the more noise is eliminated. The PSNR value increases (meaning the image quality is closer to the original image). Therefore, the averaging method is effective for reducing random noise if many frames are available.

**4. Create image masking**

```python
def image_masking():
    print("=== IMAGE MASKING ===")

    original = cv.imread('/content/drive/MyDrive/Images/couple.tiff', cv.IMREAD_GRAYSCALE)
    if original is None:
        print("Error loading image: /content/drive/MyDrive/Images/couple.tiff")
        return

    mask = np.zeros(original.shape[:2], dtype=np.uint8)
    center = (original.shape[1]//2, original.shape[0]//2)
    radius = min(original.shape[:2])//3
    cv.circle(mask, center, radius, 255, -1)

    operations = {
        'NOT (Komplemen)': cv.bitwise_not(original),
        'OR (Atau)': cv.bitwise_or(original, mask),
        'AND (Dan)': cv.bitwise_and(original, mask),
        'NAND (Not And)': cv.bitwise_not(cv.bitwise_and(original, mask)),
        'XOR (Exclusive Or)': cv.bitwise_xor(original, mask)
    }

    plt.figure(figsize=(12, 6))

    plt.subplot(2, 3, 1)
    plt.imshow(original, cmap='gray')
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(2, 3, 2)
    plt.imshow(mask, cmap='gray')
    plt.title('Mask')
    plt.axis('off')

    plot_idx = 3
    for op_name, result in operations.items():
        plt.subplot(2, 3, plot_idx)
        plt.imshow(result, cmap='gray')
        plt.title(op_name)
        plt.axis('off')
        plot_idx += 1
        if plot_idx > 6:
            break

    plt.tight_layout()
    plt.show()

image_masking()
```

1. NOT (Complement): Inverts all pixel values (black becomes white, white becomes black)
2. OR (Or): Produces a white pixel if either input is white
3. AND (And): Produces a white pixel only if both inputs are white
4. NAND (Not And): The opposite of the AND operation
5. XOR (Exclusive Or): Produces a white pixel if the inputs are different
