```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
from PIL import Image, ImageOps, ImageDraw, ImageFilter
```

```python
print('--------------------------------')
print('Mengubah tingkat kecerahan citra')
print('--------------------------------')

try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')
    exit()

original = cv.imread('/content/drive/MyDrive/Images/female.png')
brightness_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            brightness_image[y, x, c] = np.clip(original[y, x, c] + brightness, 0, 255)

final_frame = cv.hconcat((original, brightness_image))

cv2_imshow(final_frame)
```

```python
print('--------------------------------')
print('Mengubah kontras dan kecerahan citra')
print('--------------------------------')

try:
    a = float(input('Masukkan nilai kontras: '))
    b = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')
    exit()

original = cv.imread('/content/drive/MyDrive/Images/female.png')
contrast_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            contrast_image[y, x, c] = np.clip(a * original[y, x, c] + b, 0, 255)

final_frame = cv.hconcat((original, contrast_image))

cv2_imshow(final_frame)
```

```python
print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('--------------------------------------------------------')

try:
  brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

female_img = cv.imread('/content/drive/MyDrive/Images/female.png')
log_brightness = np.zeros(female_img.shape, female_img.dtype)

for r in range(female_img.shape[0]):
  for g in range(female_img.shape[1]):
    for b in range(female_img.shape[2]):
      log_brightness[r, g, b] = np.clip(brightness * np.log(1 + female_img[r, g, b]), 0, 255)

final_frame = cv.hconcat((female_img, log_brightness))
cv2_imshow(final_frame)
```
